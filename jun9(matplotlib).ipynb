```python
import matplotlib.pyplot as plt
```

```python
x=[2,3,4,5]
y=[6,7,8,9]
x1=[6,7,8,9]
y1=[2,3,4,5]
x2=[10,11,12,13]
y2=[14,15,16,17]
x3=[14,15,16,17]
y3=[10,11,12,13]
plt.plot(x,y)     #plot- plots in x,y axis
#plt.show
```

```python
plt.subplot(2,2,1)   #here 1 represents index i.e, graph number--subplot is smaller version of plot
plt.plot(x,y)

plt.subplot(2,2,2)
plt.plot(x1,y1)
```

```python
plt.subplot(2,2,3)   #here 1 represents index i.e, graph number--subplot is smaller version of plot
plt.plot(x2,y2)
plt.title('Graph-3')       #title of each graph
plt.suptitle('Data')       #super title- title for all graphs
plt.xlabel('x-axis')
plt.ylabel('y-axis')

plt.subplot(2,2,4)
plt.plot(x3,y3)
plt.title('Graph-4')
plt.suptitle('Data')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
```
